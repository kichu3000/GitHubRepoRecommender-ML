```python
import pandas as pd

sample_1 = pd.read_csv(
    "../data/split/part_1.csv",usecols=["combined_text"]
    ).sample(
        n=20000,
        random_state=42
    ) #42 beacuse it Answer to the Ultimate Question of Life, the Universe, and Everything,

sample_2 = pd.read_csv(
    "../data/split/part_2.csv",usecols=["combined_text"]
    ).sample(
        n=20000,
        random_state=42
    )

sample_3 = pd.read_csv(
    "../data/split/part_3.csv",usecols=["combined_text"]
    ).sample(
        n=20000,
        random_state=42
    )

sample_4 = pd.read_csv(
    "../data/split/part_4.csv",usecols=["combined_text"]
    ).sample(
        n=20000,
        random_state=42
    )
```

```python
sample_text = pd.concat([sample_1,sample_2,sample_3,sample_4],ignore_index=True)
```
