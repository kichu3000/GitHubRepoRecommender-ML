```python
from ast import literal_eval

df["topics"] = df["topics"].apply(literal_eval)
```

```python
df["topics"]
```

```python
df["description"]
```

```python
df.info()
```

```python
def combine(row):
    desc = row["description"]
    topics = row["topics"]
    language = row["language"]

    topics = " ".join(topics)
    

    return desc + " " + topics + " " + language
```

```python
df["combined"] = df.apply(combine, axis=1)
```

```python
df.info()
```

```python
df.to_csv("../data/combined_dataset.csv",index=False)
```
