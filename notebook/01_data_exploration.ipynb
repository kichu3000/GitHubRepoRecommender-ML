```python
import pandas as pd
df = pd.read_csv("../data/dataset_100k.csv")
```

```python
keep = ["id", "full_name", "description", "topics", "language",
        "stargazers_count", "forks_count", "created_at", "pushed_at",
        "fork", "archived"]
repos = df[keep].copy()
```

```python
repos = repos[(repos['fork'] == False) & (repos['archived'] == False)]
```

```python
repos.shape
```

```python
repos = repos.drop(columns=["fork","archived"])
```

```python
repos.shape
```

```python
repos.isna().sum()
```

```python
repos["description"] = repos["description"].fillna("")
```

```python
repos = repos[repos["description"].str.len() > 0]
```

```python
repos[repos['language'].isna()]
```

```python
repos["language"] = repos["language"].fillna("")
repos = repos[repos["language"].str.len() > 0]
```

```python
repos.shape
```

```python
repos = repos[repos["topics"] != "[]"]
```

```python
repos.shape
```

```python
repos.to_csv("../data/goodset.csv",index=False)
```
