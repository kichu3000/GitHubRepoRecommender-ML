```python
n = len(df)
chunk_size = n // 4

for i in range(4):
    start = i * chunk_size
    end = (i + 1) * chunk_size if n < 3 else n
    chunk = df.iloc[start:end]

    chunk.to_csv(
        f"../data/split/part_{i+1}.csv",index=False
    )
    print("Finished")
```
