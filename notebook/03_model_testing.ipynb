```python
import joblib
import pandas as pd
from scipy import sparse

tfidf = joblib.load("../models/tfidf.joblib")
X = sparse.load_npz("../models/tfidf_matrix.npz")
df = pd.read_csv("../data/combined_dataset.csv")
```

```python
from sklearn.metrics.pairwise import linear_kernel
def recommend(user_input,count):
    user_query = tfidf.transform([user_input])
    score = linear_kernel(user_query,X).flatten()
    top = score.argsort()[::-1][:count]
    results = df.iloc[top][["id","full_name","combined"]]
    return results
```
